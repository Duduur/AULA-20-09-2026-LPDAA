```python
limite = 8000

try:
    valor_saque = float(input("Digite o valor do saque: "))
    parcelas = int(input("Digite o número de parcelas: "))

    limite_restante = limite - valor_saque

    if valor_saque < 100 or valor_saque > 5000:
        print(f"Valor saque solicitado: {valor_saque}")
        print("Valor do saque inválido.")

    elif limite_restante < 3500:
        print(f"Limite restante: {limite_restante}")    
        print("Limite insuficiente para realizar o saque.")

    elif parcelas < 1 or parcelas > 6:
        print("Número de parcelas não permitido.")

    else:
        valor_parcela = valor_saque / parcelas

        if valor_parcela < 50:
            print(f"Valor da parcela: {valor_parcela:.2f}")
            print("Valor da parcela não permitido.")

        else:
            limite = limite_restante
            print(f"O valor de cada parcela é: {valor_parcela:.2f}")

    if limite != limite_restante:
        raise ValueError

except ValueError:
    print("Saque não autorizado ou valor inválido!")

except ZeroDivisionError:
    print("O número de parcelas não pode ser zero.")

else:
    print("Saque realizado com sucesso!")

finally:
    print("Operação finalizada.")
```

```python
limite = 8000
taxa = 0.02

try:
    valor_saque = float(input("Digite o valor do saque: "))
    parcelas = int(input("Digite o número de parcelas: "))
    escolher_juros = input("Escolha 1 para Juros Simples ou 2 para Juros Compostos: ")

    limite_restante = limite - valor_saque

    if valor_saque < 100 or valor_saque > 5000:
        print("Valor do saque inválido.")

    elif limite_restante < 3500:
        print("Limite insuficiente para realizar o saque.")

    elif parcelas < 1 or parcelas > 6:
        print("Número de parcelas não permitido.")

    elif taxa <= 0:
        print("Taxa de juros inválida.")

    elif escolher_juros != "1" and escolher_juros != "2":
        print("Tipo de juros inválido.")

    else:
        if escolher_juros == "1":
            montante = valor_saque * (1 + taxa * parcelas)
            print("Juros Simples selecionado.")

        else:
            montante = valor_saque * (1 + taxa) ** parcelas
            print("Juros Compostos selecionado.")

        valor_parcela = montante / parcelas

        if valor_parcela < 50:
            print(f"Valor da parcela: R$ {valor_parcela:.2f}")
            print("Valor da parcela não permitido.")

        else:
            limite = limite_restante
            print(f"Valor total com juros: R$ {montante:.2f}")
            print(f"Valor de cada parcela: R$ {valor_parcela:.2f}")
            print(f"Limite disponível: R$ {limite:.2f}")

    if limite != limite_restante:
        raise ValueError

except ValueError:
    print("Saque não autorizado ou valor inválido!")

else:
    print("Saque realizado com sucesso!")

finally:
    print("Operação finalizada.")
```
