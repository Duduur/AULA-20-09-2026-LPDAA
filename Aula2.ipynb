```python
def registrar_ocorrencia(codigo, *sistemas, **informacoes):
    print(f"O codigo de ocorrência é: {codigo}")

    print(f"Sistemas envolvidos:{len(sistemas)}")
    for i, sistema in enumerate(sistemas, start=1):
        print(f"{i} - {sistema}")

    print("Informações adicionais:")
    for chave, valor in informacoes.items():
        print(f"- {chave}: {valor}")
```

```python
def analise_vendas(vendedor, *vendas):
    print(f"Desempenho do vendedor: {vendedor}, quantidade de vendas: {len(vendas)}")
    total = sum(vendas)
    print(f"Total de vendas: {total}")
    print(f"Média de vendas: {sum(vendas) / len(vendas):.2f}")
    print(f"Maior valora de venda: {max(vendas)}")

    if total >= 5000:
        print("Parabéns! Você atingiu a meta de vendas!")
    else:
        print("Você não atingiu a meta de vendas. Continue se esforçando!")
        
```

```python
print(analise_vendas("João", 1000, 2000, 3000, 4000, 5000))
print(analise_vendas("Maria", 1500, 2500, 3500))
print(analise_vendas("Pedro", 500, 1000, 1500, 2000, 2500, 3000))
```

```python
def registrar_atendimento(cliente, *procedimentos, **informacoes):
    print(f"Cliente: {cliente}")
    print(f"Quantidade de procedimentos: {(procedimentos)}")
    for i, procedimento in enumerate(procedimentos, start=1):
        print(f"{i} - {procedimento}")

    print("Informações adicionais:")
    for chave, valor in informacoes.items():
        print(f"- {chave}: {valor}")

    if "data_retorno" in informacoes:
        print(f"Data de retorno: {informacoes['data_retorno']}")
    else:
        print("Retorno não informado")
```

```python
setores = ["Financeiro", "Recursos Humanos", "Comercial"]

def cadastrar_funcionario(nome, idade, salario, setor):
    if nome == "":
        print("Nome do funcionário não pode ser vazio.")
        return
    elif idade < 18 or idade > 70:
        print("Idade inválida para cadastro.")
        return
    elif salario < 0:
        print("Salário não pode ser negativo.")
        return
    elif setor not in setores:
        print("Setor inválido. Setores válidos: Financeiro, Recursos Humanos, Comercial.")
        return

    print(f"Funcionário cadastrado com sucesso: {nome}, {idade} anos, salário R$ {salario:.2f}, setor {setor}.")
    
```

```python
print(cadastrar_funcionario("João", 0, 0, "Financeiro"))
```
