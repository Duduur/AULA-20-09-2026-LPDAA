```python
tempos = [48, 32, 55, 41, 37, 62, 29, 45, 51, 35, 70]


def calcular_media(lista):
    soma = 0
    for valor in lista:
        soma = soma + valor
    return soma / len(lista)


def calcular_mediana(lista):
    ordenados = sorted(lista)
    meio = len(ordenados) // 2
    if len(ordenados) % 2 == 1:
        return ordenados[meio]
    return (ordenados[meio - 1] + ordenados[meio]) / 2


def calcular_quartis(lista):
    ordenados = sorted(lista)
    meio = len(ordenados) // 2
    metade_de_baixo = ordenados[:meio]
    if len(ordenados) % 2 == 1:
        metade_de_cima = ordenados[meio + 1:]
    else:
        metade_de_cima = ordenados[meio:]
    q1 = calcular_mediana(metade_de_baixo)
    q2 = calcular_mediana(ordenados)
    q3 = calcular_mediana(metade_de_cima)
    return q1, q2, q3


def calcular_amplitude(lista):
    return max(lista) - min(lista)


def calcular_variancia(lista):
    media = calcular_media(lista)
    soma_quadrados = 0
    for valor in lista:
        diferenca = valor - media
        soma_quadrados = soma_quadrados + diferenca ** 2
    return soma_quadrados / (len(lista) - 1)


def calcular_desvio_padrao(lista):
    return calcular_variancia(lista) ** 0.5


def calcular_iqr(lista):
    q1, q2, q3 = calcular_quartis(lista)
    return q3 - q1


def contar_perto_da_media(lista):
    media = calcular_media(lista)
    desvio = calcular_desvio_padrao(lista)
    limite_baixo = media - desvio
    limite_alto = media + desvio
    perto = 0
    for valor in lista:
        if valor >= limite_baixo and valor <= limite_alto:
            perto = perto + 1
    return perto


q1, q2, q3 = calcular_quartis(tempos)

print("Tempo Médio (minutos):", round(calcular_media(tempos), 2))
print("Tempo Mediano (minutos  ):", calcular_mediana(tempos))
print("Q1:", q1, "Q2:", q2, "Q3:", q3)
print("Amplitude:", calcular_amplitude(tempos))
print("Variância:", round(calcular_variancia(tempos), 2))
print("Desvio padrão:", round(calcular_desvio_padrao(tempos), 2))
print("IQR:", calcular_iqr(tempos))
print("Perto da média:", contar_perto_da_media(tempos), "de", len(tempos))
```

```python
from datetime import datetime

data_pedido = "12/09/2026"
data_entrega = "18/09/2026"

data_pedido_t = datetime.strptime(data_pedido, "%d/%m/%Y")
data_entrega_t = datetime.strptime(data_entrega, "%d/%m/%Y")

diferenca = data_entrega_t - data_pedido_t

if diferenca.days > 5:
    print("A entrega está atrasada")
else:
    print("Entrega dentro do prazo")

print(f"Data do pedido: {data_pedido_t.strftime('%d/%m/%Y')}")
print(f"Data de entrega: {data_entrega_t.strftime('%d/%m/%Y')}")
print(f"Dias até ser entregue: {diferenca.days}")

```
